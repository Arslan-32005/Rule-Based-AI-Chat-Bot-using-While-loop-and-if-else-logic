A rule-based AI chatbot built with Python using if-else logic and while loops. Responds to greetings, questions, and exit commands with predefined responses.


```python
print("Welcome to Rule-Based AI Chatbot!")
print("Type 'help' to see available commands")
print("Type 'bye' to exit\n")
```

```python
# Start the continuous loop
while True:
    # Get user input from user
    user_input = input("You: ")
    
    # Convert to lowercase for easy matching
    user_input = user_input.lower()
    
    # Now we check what the user said using if-else
    # Check if user wants to exit
    if user_input == "bye" or user_input == "exit" or user_input == "quit":
        print("Bot: Goodbye! See you later!")
        break  
    # Check if user is greeting
    elif user_input == "hello" or user_input == "hi" or user_input == "hey":
        print("Bot: Hello! How can I help you?")  
            # Check if user asks how we are
    elif user_input == "how are you":
        print("Bot: I am doing great, thanks for asking!")
            # Check if user asks how we are
    elif user_input == "how are you":
        print("Bot: I am doing great, thanks for asking!")
            # Check if user asks our name
    elif user_input == "what is your name" or user_input == "who are you":
        print("Bot: I am a Rule-Based Chatbot!")
            # Check if user asks for help
    elif user_input == "help":
        print("\nAvailable commands:")
        print("- hello/hi/hey : Get a greeting")
        print("- how are you : Ask how I am")
        print("- what is your name : Know my name")
        print("- bye/exit/quit : Exit chatbot\n")
            # If user says something we don't recognize
    else:
        print("Bot: I don't understand that. Try 'help' for commands.")    
```
